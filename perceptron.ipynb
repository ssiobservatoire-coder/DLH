```python
def plot_dataset(X, y, title):

    plt.figure(figsize=(6, 5))
    plt.scatter(X[0,0], X[0,1], marker='^') #first class
    plt.scatter(X[1:,0], X[1:,1], marker='o') #second class


    plt.title(title)
    plt.grid(True)
    plt.show()


plot_dataset(X_or, y_or, "Perceptron Decision Boundary (OR)")
```
